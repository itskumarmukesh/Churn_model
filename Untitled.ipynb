```python
import sys
sys.path.append("C:/Users/slmkc/Downloads/PhaseI/Churn_model/src")
from churn.config import load_config
from churn.data import load_and_prepare, make_split
from churn.model import build_models
from churn.evaluate import out_of_fold_probs, choose_threshold

cfg = load_config("C:/Users/slmkc/Downloads/PhaseI/Churn_model/configs/telco.yaml")
X, y, ids = load_and_prepare(cfg)
X_train, X_test, y_train, y_test = make_split(X, y, cfg)
models = build_models(X_train, cfg["random_state"])

oof = out_of_fold_probs(models["Logistic Regression"], X_train, y_train, cfg)
print("Average predicted probability:", round(oof.mean(), 3))
threshold, table = choose_threshold(y_train.values, oof, cfg["business"])
print("Chosen threshold:", threshold)
```
